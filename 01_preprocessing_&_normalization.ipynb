1. Imports AND Load Expression Matrix


```python
import pandas as pd
import numpy as np

df = pd.read_csv("expression_matrix.csv")
df.head()
```

2. Log₂ Normalization

```python
ids = df.iloc[:, 0]
expr = df.iloc[:, 1:]

log2_expr = np.log2(expr + 1)

log2_df = pd.concat([ids, log2_expr], axis=1)
log2_df.head()
```

3. Save Output

```python
log2_df.to_csv("expression_matrix_log2.csv", index=False)
```
